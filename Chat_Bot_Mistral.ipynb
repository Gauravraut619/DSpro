```python
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig, pipeline
import torch

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_use_double_quant=True,
)
```

```python

model = "mistralai/Mistral-7B-Instruct-v0.1"

tokenizer = AutoTokenizer.from_pretrained(model, use_auth_token=True)
```

```python
from transformers import pipeline

mistral_pipeline = pipeline(
    "text-generation",  # LLM task
    model=model,
    torch_dtype=torch.float16,
    device_map="auto",
)
```

```python
def get_mistral_response(prompt: str) -> None:
    """
    Generate a response from the mistral model.

    Parameters:
        prompt (str): The user's input/question for the model.

    Returns:
        None: Prints the model's response.
    """
    sequences = mistral_pipeline(
        prompt,
        do_sample=True,
        top_k=10,
        num_return_sequences=1,
        eos_token_id=tokenizer.eos_token_id,
        max_length=256,
    )
    print("Chatbot:", sequences[0]['generated_text'])



prompt = 'tell me the  discoveries of newton?\n'
get_mistral_response(prompt)
```

```python
prompt = 'who is president of USA?\n'
get_mistral_response(prompt)
```

```python
prompt = 'what is D.O.B OF president?\n'
get_mistral_response(prompt)
```

Conversationl Chat Bot

```python
while True:
    user_input = input("You: ")
    if user_input.lower() in ["bye", "quit", "exit"]:
        print("Chatbot: Goodbye!")
        break
    get_mistral_response(user_input)
```
